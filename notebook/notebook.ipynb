```python
import pandas as pd 
import numpy as np
df=pd.read_csv('superstore_cleaned.csv')
df.head()
```

```python
q1=df['Sales'].quantile(0.25)
q3=df['Sales'].quantile(0.75)
iqr=q3-q1
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr  
outliers=df[(df['Sales'] < lower_bound) | (df['Sales'] > upper_bound)]

print("Number of sales outliers:", len(outliers))
print(outliers[["Order ID", "Product Name", "Region", "Sales", "Profit"]].head(10))

outliers.to_csv("sales_outliers.csv", index=False)
```

```python
monthly_profit = df.groupby("Month_Year", as_index=False)["Profit"].sum()
monthly_profit["Profit_Change"] = monthly_profit["Profit"].diff()

print(monthly_profit)

monthly_profit.to_csv("monthly_profit_trend.csv", index=False)
```
